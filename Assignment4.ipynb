```python
import pandas as pd
```

```python
#Q1
roll_number = input("Enter your roll number: ")
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python
# Take last two digits of roll number
last_digit1 = int(roll_number[-2])
last_digit2 = int(roll_number[-1])

categories = ["billing", "account", "general"]

personalized_entries = []

# First personalized entry
d = last_digit1
category = categories[d % 3]

if category == "billing":
    entry1 = {
        "question": "how can i get my payment receipt",
        "answer": "You can download it from the payment section.",
        "keywords": "payment receipt fee",
        "category": "billing"
    }

elif category == "account":
    entry1 = {
        "question": "how do i update my mobile number",
        "answer": "Go to account settings to update it.",
        "keywords": "mobile update account",
        "category": "account"
    }

else:
    entry1 = {
        "question": "how can i contact support",
        "answer": "You can contact support by email.",
        "keywords": "contact support help",
        "category": "general"
    }

personalized_entries.append(entry1)


# Second personalized entry
d = last_digit2
category = categories[d % 3]

if category == "billing":
    entry2 = {
        "question": "how can i check my payment status",
        "answer": "Check it in the payment section.",
        "keywords": "payment status fee",
        "category": "billing"
    }

elif category == "account":
    entry2 = {
        "question": "how do i change my password",
        "answer": "Go to account settings to change it.",
        "keywords": "password change account",
        "category": "account"
    }

else:
    entry2 = {
        "question": "where can i get help",
        "answer": "You can contact customer support.",
        "keywords": "help support customer",
        "category": "general"
    }

personalized_entries.append(entry2)


# Combine fixed and personalized entries
all_entries = fixed_entries + personalized_entries

# Create DataFrame
df = pd.DataFrame(all_entries)

print(df)
```

```python
# Q2

def score_query(query, df):

    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():

        score = 0

        # Combine question and keywords
        text = row["question"].lower() + " " + row["keywords"].lower()

        # Check every word in the query
        for word in query_words:

            if word in text:
                score = score + 1

        # Add only matching entries
        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "score": score
            })

    # Convert to DataFrame
    result_df = pd.DataFrame(results)

    # Sort by score
    result_df = result_df.sort_values(
        by="score",
        ascending=False
    )

    return result_df


# Take query from user
query = input("Enter your query: ")

# Call function
result = score_query(query, df)

# Print result
if result.empty:
    print("No matching entry found")

else:
    print(result)
```

```python
# Q3

def same_category(category_name, df):

    result = df[df["category"] == category_name]

    return result["question"]


category = input("Enter category: ")

print(same_category(category, df))
```

```python
# Q4

entry_number = int(input("Enter entry number: "))

new_keyword = input("Enter new keyword: ")

# Add keyword
df.loc[entry_number, "keywords"] = (
    df.loc[entry_number, "keywords"] + " " + new_keyword
)

# Save DataFrame as CSV
file_name = roll_number + "_faq_data.csv"

df.to_csv(file_name, index=False)

print("File saved successfully")
```

```python
# Q5

result = df.groupby("category").size()

print(result)
```

```python
# Q6

def best_matches(query, df):

    result = score_query(query, df)

    # If no match found
    if result.empty:
        print("No matching entry found")

    else:
        # Find highest score
        highest_score = result["score"].max()

        # Print all entries with highest score
        best = result[result["score"] == highest_score]

        print(best)


# Example with tie
best_matches("fee", df)

# Example without tie
best_matches("password", df)
```
