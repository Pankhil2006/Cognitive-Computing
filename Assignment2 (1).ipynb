```python

# Q1

roll_no = int(input("Enter your roll number: "))

L = []
while(roll_no):
    digit = roll_no%10;
    L.append(digit*10);
    roll_no = int(roll_no/10)

#i
print("L =", L) 
#ii
L.append(50)
print("After append:", L)     
L.insert(2, 90)
print("After insert:", L)   
#iii
L.remove(20)
print("After remove:", L)

L.pop()
print("After pop:", L)
#iv
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

#v
print("First three:", L[:3])
print("Last three:", L[-3:])
#vi
average = sum(L) / len(L)
new_list = []
for x in L:
    if x>average:
        new_list.append(x)
print("Average =", average)
print("Elements greater than average:", new_list)
```

```python
# Q2
scores = tuple(L[:8])
print("Scores:", scores)
#i
highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)

print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)
#ii
reversed_scores = list(scores[::-1])

print("Reversed list:", reversed_scores)

#iii
score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score is not present in the tuple.")

#v
first, second, *remaining = scores

print("First score:", first)
print("Second score:", second)
print("Remaining scores:", remaining)
```

```python
# Q3

import random

random.seed(1024170322)

numbers = []
#i
for i in range(0,100):
    number = random.randint(100, 900)
    numbers.append(number)

print(numbers)

#ii
odd_count = 0

for number in numbers:
    if number % 2 != 0:
        odd_count += 1

print("Number of odd numbers:", odd_count)

#iii
even_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count += 1

print("Number of even numbers:", even_count)

#iv
prime_numbers = []

for number in numbers:
    if number > 1:
        is_prime = True

        for i in range(2, number):
            if number % i == 0:
                is_prime = False
                break

        if is_prime:
            prime_numbers.append(number)

print("Number of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)

#v
most_frequent = numbers[0]
highest_count = numbers.count(numbers[0])

for number in numbers:
    count = numbers.count(number)

    if count > highest_count:
        highest_count = count
        most_frequent = number

print("Most frequent number:", most_frequent)
print("It occurs", highest_count, "times")
```

```python
# Q4

digits = []
roll_number = input("Enter your roll number: ")
roll= int(roll_number[0:8])
while(roll):
    digit = roll%10;
    digits.append(digit);
    roll = int(roll/10)

print(digits)
A = set()
B = set()

for digit in digits:
    A.add(digit * 7)
    B.add(digit * 9)

print("A =", A)
print("B =", B)
#vi
union_set = A.union(B)

print("Union:", union_set)
#vii
intersection_set = A.intersection(B)

print("Intersection:", intersection_set)
#viii
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)
#ix
symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)
#x
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
#xi
X = int(input("Enter a value to remove from A: "))

A.discard(X)

print("A after discard:", A)
# discard() does not give an error if the value is not present in the set.
```

```python
# Q5

my_dict = {
    "name": "Pankhil Gupta",
    "roll_no": "1024170322",
    "branch": "COPC",
    "age": 19,
    "city": "Ludhiana"
}

print(my_dict)
#i
value = my_dict["city"] 
my_dict.pop("city")
my_dict["location"] = value
print(my_dict)
#ii
my_dict["cgpa"] = 9.7
print(my_dict)
#iii
my_dict["age"] = my_dict["age"] + 1
print(my_dict)
#iv
dict1 = my_dict.copy()
dict2 = my_dict.copy()
dict1.pop("branch")

print("After pop:", dict1)

del dict2["branch"]

print("After del:", dict2)
# pop() removes the key and returns its value, while del only removes the key-value pair.

#v
for key, value in my_dict.items():
    print(key, "→", value)
#vi
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email is not present in the dictionary.")

#vii
friend_dict = {
    "name": "ABC",
    "roll_no": "10240000",
    "branch": "COPC",
    "age": 19,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:", merged_dict)
# If both dictionaries have the same key, the value from friend_dict wins.
#viii
string_values = {}

for key, value in my_dict.items():
    if type(value) == str:
        string_values[key] = value

print("String values:", string_values)
```
